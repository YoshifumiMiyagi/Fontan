# 02 — Multiple imputation + fixed-reference Nyström projection
Run 20 stochastic imputations while keeping the reference manifold fixed.

```python
import pandas as pd
from src.fontan_manifold import FontanManifold
```

```python
DATA_PATH = '/content/drive/MyDrive/Fontan/your_data.csv'
raw = pd.read_csv(DATA_PATH)
df7 = FontanManifold.prepare_dataframe(raw)

# Rebuild the same locked reference deterministically
model = FontanManifold(n_neighbors=50, n_dm=5, id_col='subj_id')
model.fit(df7)
```

## 20 imputations
The diffusion reference is not refitted inside this loop.

```python
projection_long, mi_diagnostics = model.project_multiple_imputation(
    df7,
    n_imputations=20,
    seeds=list(range(1001, 1021)),
    n_dm=3,
)

projection_summary = model.summarize_mi(projection_long, n_dm=3)
display(mi_diagnostics)
display(projection_summary.head())
```
