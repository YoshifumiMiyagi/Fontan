# 03 — Outcome analysis
Fit adjusted logistic models within each imputation and pool the diffusion-component coefficients using Rubin's rules.

```python
import pandas as pd
from src.fontan_manifold import FontanOutcomeAnalyzer, default_sex_binary
```

```python
DATA_PATH = '/content/drive/MyDrive/Fontan/your_data.csv'
PROJECTION_PATH = 'projection_long.csv'

clinical = pd.read_csv(DATA_PATH)
projection_long = pd.read_csv(PROJECTION_PATH)

# Change `sex` below if the source column has another name.
clinical['sex_binary'] = default_sex_binary(clinical['sex'])
```

## DM-specific adjusted models
Default model: outcome ~ DM_z + fontan_year + sex_binary.

```python
analyzer = FontanOutcomeAnalyzer(id_col='subj_id')

mi_results = analyzer.fit_univariable_dm_models(
    projection_long=projection_long,
    clinical_df=clinical,
    dms=('DM1', 'DM2', 'DM3'),
    covariates=('fontan_year', 'sex_binary'),
)

pooled = analyzer.pool(mi_results, fdr=True)
display(pooled[['DM','Outcome','N','Events','OR_95CI','P_value','q_value','FMI']])
```

## Export manuscript-ready results

```python
pooled.to_csv('pooled_outcome_results.csv', index=False)
```
